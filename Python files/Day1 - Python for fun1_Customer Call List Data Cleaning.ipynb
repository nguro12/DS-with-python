# DATA CLEANING

### dealing with missing values

```python
# what is the distribution of missing values per column?
df.isna().sum()
```

```python
# Better yet, let's view the proportion of missing values per column
(df.isna().mean()*100).round().sort_values(ascending= False)
```

```python
# FILLING MISSING VALUES

#df = df.dropna()# Drop Rows with Missing Values
#df = df.dropna(subset=['Do_Not_Contact', 'Phone_Number', 'Last_Name') #Drop rows with missing values in specific columns 

dncMode = df['Do_Not_Contact'].mode()[0]
df['Do_Not_Contact'] = df['Do_Not_Contact'].fillna(dncMode)
#df['Do_Not_Contact'].isna().sum()
df['Last_Name']= df['Last_Name'].fillna('.')
df['Phone_Number'] = df['Phone_Number'].fillna('000')
df.isna().sum()
```

## Standardizing Variable values

```python
for col in df.select_dtypes(include='object').columns:
    print(f"'{col}': {df[col].unique()}\n")
```

```python
df['Paying Customer']= df['Paying Customer'].map( {'Yes': 'Yes',  
                                               'Y': 'Yes',
                                               'No': 'No',
                                               'N': 'No',
                                               'N/a':' '} )

df['Paying Customer']= df['Paying Customer'].fillna(df['Paying Customer'].mode()[0]) #fill missing variables
df['Paying Customer'].unique()
```

```python
df['Do_Not_Contact']= df['Do_Not_Contact'].map({'Yes':'Yes',
                                               'Y':'Yes',
                                               'No':'No',
                                               'N':'No',})
df['Do_Not_Contact'].unique()
```

```python
# Standardize only one value and leave others unchanged
df['Last_Name'] = df['Last_Name'].replace({'...Potter': 'Potter'})
```
