```python
#dealing with missing values
#df= df["Price"].dropna() --gives df as a series 
df= df[df["Price"].notna()]
```

```python
df.isna().sum()
```

```python
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
```

```python
df.columns
```

```python
x= df.drop("Price", axis=1)
y = df["Price"]

x.shape, y.shape
```

```python
y.isna().sum(), x.isna().sum()
```

```python
df.isna().sum()
```

```python
x['Make'].mode()
```

```python
#how to fill/impute
first_impute = SimpleImputer(strategy="constant", fill_value=4)
second_impute = SimpleImputer(strategy="most_frequent")
third_impute = SimpleImputer(strategy='mean')

#what to fill/impute
first_fill= ["Doors"]
second_fill = ["Make", "Colour"]
third_fill = ["Odometer (KM)"]

# blueprint/model for filing/imputing
blueprint = ColumnTransformer(transformers= [("first_impute", first_impute, first_fill),
                   ("second_impute", second_impute, second_fill),
                   ("third_impute", third_impute, third_fill)
                  ], remainder= "passthrough")
blueprint

#preserve the pandas output ie. original column names
#blueprint.set_output(transform="pandas")

#fitting the blueprint/model to fill the data

filled_x = blueprint.fit_transform(x)
filled_x
```

```python
filled_x= pd.DataFrame(filled_x, columns= x.columns)
```
