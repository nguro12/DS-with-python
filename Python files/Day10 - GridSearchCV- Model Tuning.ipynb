```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.read_csv("C:\\Users\\User\\Desktop\\Python files\\data\\car-sales-extended.csv")
df.head()
```

```python
# Splitting the data
x = df.drop("Price", axis=1)
y = df["Price"]

x.shape, y.shape
```

## Encode the data

```python
#columns to transform
cat_cols= ['Make', 'Colour', 'Doors']

#transfomer 
transformer= ColumnTransformer( transformers=  [('one_hot', OneHotEncoder(sparse_output= False
                                                                ,handle_unknown='ignore'), cat_cols)],
                               remainder= "passthrough")
transformer;
```

```python
t_x= transformer.fit_transform(x)
t_x;
```

```python
t_x= pd.DataFrame(t_x)
t_x.head(10)
```

```python
# looking into how encoding is done
print(x['Make'].unique())

print(x['Colour'].unique())

print(x['Doors'].unique())

t_x.shape
```

## Model Fitting

```python
from sklearn.metrics import accuracy_score
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
```

```python
#split the data
x_train, x_test, y_train, y_test = train_test_split(t_x,y, test_size= 0.2)
x_train.shape, x_test.shape, y_train.shape, y_test.shape 
```

```python
#train the model
model= RandomForestRegressor().fit(x_train, y_train)
```

## Model Tuning

Manual Model Tuning

```python
model.get_params()
```

```python
# Checking the model perfomance on the training data
print(f"The accuracy score of this model in the training data is: {model.score(x_train, y_train)*100:.2f}%")
```

```python
#tesing the model on the test data
model.predict(x_test)
# Checking the model perfomance on the test data
print(f"The accuracy of this model on the test data is: {model.score(x_test, y_test)*100:.2f}%")
```

```python
#manually checking the model's perfomance accross adjusted parameters

for i in range(100, 300, 15):
    model2 = RandomForestRegressor(n_estimators= i).fit(x_train, y_train)
    print(f"The accuracy score of our model with {i} n_estimatiors is:\
    {model2.score(x_test, y_test)*100:.2f}%")
```

## GridSearchCV

Automated Model Tuning 

```python
#Define the base model
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(
    random_state=42,
    n_jobs=-1 #n_jobs=-1 Controls how many CPU cores are used during training and prediction.(-1 means use all available processor cores.)
)
rf.get_params()
```

```python
#Step 2: Define the parameter grid
param_grid = {
    "n_estimators": [10, 25, 50, 75, 100],
    "max_depth": [None, 5, 10, 20],
    "min_samples_split": [2, 5, 10]
}
```

```python
#Step 3: Create the GridSearchCV object
from sklearn.model_selection import GridSearchCV

grid_search = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    scoring="r2",
    cv=5,
    n_jobs=-1,
    verbose=1
)
```

```python
#Step 4: Fit GridSearchCV
grid_search.fit(x_train, y_train)
```

```python
#5. Accessing the results
#Best parameters
grid_search.best_params_
#Best cross-validated score
grid_search.best_score_
#Best trained model
best_model = grid_search.best_estimator_
```

```python
#6. Evaluate on the test set (critical step)
test_r2 = best_model.score(x_test, y_test)
print(f"Test R²: {test_r2:.3f}")
```
