# Modelling

```python
x = df.drop(columns=['R'])
y = df['R']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y,
                                                    test_size= 0.2, 
                                                    stratify= y, 
                                                  random_state= 42 )
x_train.shape, x_test.shape, y_train.shape, y_test.shape, y.value_counts(), y_train.value_counts(), y_test.value_counts()
```

```python
# fittitng the model
from sklearn.linear_model import LogisticRegression
model= LogisticRegression()
model.fit(x_train, y_train);
```

### model evaluation

```python
from sklearn.metrics import accuracy_score
# testing the data on the training data 
x_train_eval= model.predict(x_train)
x_train_eval_results= accuracy_score(x_train_eval, y_train)

print('Model\'s accuracy on training data: ',  x_train_eval_results)
```

```python
# testing the data on the testing data 
x_test_eval= model.predict(x_test);
# checkig the accuracy of the model on the testing data
x_test_eval_results= accuracy_score(y_test, x_test_eval)

print('Model\'s accuracy on test data: ', x_test_eval_results)
```
