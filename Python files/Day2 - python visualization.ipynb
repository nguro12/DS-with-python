## compare the population  of the US and China

```python
# subset us and china data
print(data['country'].value_counts())
```

```python
us = data[data.country == 'United States']
china = data[data.country == 'China']
```

```python
plt.plot(us.year, us.population);
```

```python
plt.plot(china.year, china.population);
```

```python
plt.plot(us.year, us.population / 10**6)
plt.plot(china.year, china.population/ 10**6)
plt.legend(['United States','china'])
plt.xlabel('Year')
plt.ylabel('Population in Millions')
plt.title('Comparison of US and China population over the years')
plt.show()
```

```python
# compare china and us relative population growth
plt.plot(us.year,(us.population/us.population.iloc[0])*100)
plt.plot(china.year, (china.population/china.population.iloc[0])*100)
plt.title('Comparison of US and China relative population growth over the years')
plt.xlabel('Year')
plt.ylabel('population growth (first year = 100)')
plt.legend(['United States','china'])
plt.show()
```

```python
df.head()
```

```python
df["column_d"] = ["Mark", "victor", "Antony", "Risper", "Brenda"]
```

```python
df.columns
```

```python
plt.figure(figsize= (4,3))

plt.plot(df.column_a, df.column_b, linewidth= 3,
         color= 'salmon' ,marker = '*', linestyle = '--',
         markersize= 10, markeredgecolor= 'blue')
plt.grid()
plt.title('Advanced graph practice', fontdict= {'fontsize': 23, 
                                                'fontname': 'Times New Roman'})
plt.xlabel('column_a')
plt.ylabel('column_b')
plt.legend(["disp"])
plt.show()
```

```python
# Bar Chart
plt.figure(figsize=(4,3))
plt.bar(df.column_a, df.column_b, color= ('purple', 'violet', 'indigo', 'yellow', 'orange'))
plt.show()
```

```python
#advanced Bar chart
plt.figure(figsize= (4,3))
bar= plt.bar(df.column_a, df.column_b)

patterns= [ '/', 'o', '*', '-', '.']
for b, p in zip(bar, patterns):
    b.set_hatch(p)
    
# bar[0].set_hatch('/')
# bar[1].set_hatch('o')
# bar[2].set_hatch('*')
# bar[3].set_hatch('-')
# bar[4].set_hatch('.')


plt.title('advanced Bar chart')
plt.xlabel('column_a')
plt.ylabel('column_b')
plt.savefig('#advanced Bar chart', dpi= 300)
plt.show()
```

```python
# Bar Chart
plt.figure(figsize=(4,3))
plt.barh(df.column_a, df.column_b, color = 'green')
plt.xlabel('column_b')
plt.ylabel('column_a')
plt.show()
```
